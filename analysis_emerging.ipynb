```python
warnings.filterwarnings('ignore')
# Install packages
import pandas as pd
import numpy as np
import warnings
import matplotlib.pyplot as plt
import seaborn as sns

# EDA
from data_profiling import ProfileReport

# Models



import random
import os

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)
```

```python
#Read the datasets
from pathlib import Path

data_dir = Path("data_emerging")
if not data_dir.is_dir():
    raise FileNotFoundError(f"Data directory not found: {data_dir.resolve()}")

parquet_files = sorted(data_dir.glob("*.parquet"))
if not parquet_files:
    raise FileNotFoundError(f"No Parquet datasets found in {data_dir.resolve()}")

datasets = {path.stem: pd.read_parquet(path) for path in parquet_files}

display(pd.DataFrame(
    [
        {"dataset": name, "rows": len(df), "columns": len(df.columns), "duplicate rows": df.duplicated().sum()}
        for name, df in datasets.items()
    ]
))

for name, df in datasets.items():
    print(f"\nFirst 10 rows of {name}:")
    print(df.head(10).to_string(index=False))
```

```python
# Directory for HTML reports
report_dir = Path("profile_reports_emerging")
report_dir.mkdir(parents=True, exist_ok=True)

# Datasets to profile
profile_datasets = {
    "policies_motor_emerging": datasets["policies_motor"],
    "policies_buildings_emerging": datasets["policies_buildings"],
    "policies_contents_emerging": datasets["policies_contents"],
    "addresses_emerging": datasets["addresses"],
    "customers_emerging": datasets["customers"],
}

# Generate and save reports
for name, df in profile_datasets.items():
    print(f"Generating profile report: {name}")

    profile = ProfileReport(
        df,
        title=f"Profiling Report Emerging- {name}",
        explorative=True,
        minimal=True
    )

    output_path = report_dir / f"{name}_profile_emerging.html"
    profile.to_file(output_path)

    print(f"Saved: {output_path}")
```
